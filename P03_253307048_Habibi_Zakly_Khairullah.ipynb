| &nbsp;&nbsp;&nbsp; `p` &nbsp;&nbsp;&nbsp; | &nbsp;&nbsp;&nbsp; `q` &nbsp;&nbsp;&nbsp; | &nbsp;&nbsp;&nbsp; `p ∧ q` &nbsp;&nbsp;&nbsp; | &nbsp;&nbsp; `¬(p ∧ q)` &nbsp;&nbsp; | &nbsp;&nbsp;&nbsp; `¬p` &nbsp;&nbsp;&nbsp; | &nbsp;&nbsp;&nbsp; `¬q` &nbsp;&nbsp;&nbsp; | &nbsp;&nbsp; `¬p ∨ ¬q` &nbsp;&nbsp; |
| :---: | :---: | :---: | :---: | :---: | :---: | :---: |
| B | B | B | **S** | S | S | **S** |
| B | S | S | **B** | S | B | **B** |
| S | B | S | **B** | B | S | **B** |
| S | S | S | **B** | B | B | **B** |

Kegiatan 1

```python
import numpy as np                    # np = nama panggilan alias untuk library NumPy
uts = np.array([78, 90, 66, 85, 72])  # list -> array NumPy
uas = np.array([80, 94, 52, 88, 75])

nilai_akhir = 0.4 * uts + 0.6 * uas   # rumus dihitung sekaligus untuk 5 mahasiswa
print ("Nilai akhir   : ", nilai_akhir)
print ("Rata-rata     : ", nilai_akhir.mean())
print ("Tertinggi     : ", nilai_akhir.max())
```

Pengembangan 1

```python
import numpy as np  # np = nama panggilan alias untuk library NumPy

# Data jumlah jajan selama 5 hari (Senin - Jumat)
kopi_per_hari = np.array([2, 1, 3, 0, 2])  # jumlah gelas kopi yang dibeli
roti_per_hari = np.array([1, 2, 1, 1, 3])  # jumlah bungkus roti yang dibeli

# Harga satuan (dalam Rupiah)
harga_kopi = 15000
harga_roti = 10000

# Rumus dihitung sekaligus untuk pengeluaran 5 hari (operasi vektor)
pengeluaran_harian = (harga_kopi * kopi_per_hari) + (harga_roti * roti_per_hari)

# Menampilkan hasil
print("Pengeluaran harian (Rp) : ", pengeluaran_harian)
print("Total jajan seminggu    : ", pengeluaran_harian.sum())
print("Rata-rata jajan per hari: ", pengeluaran_harian.mean())
print("Pengeluaran tertinggi   : ", pengeluaran_harian.max())
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Muat data
df = pd.read_csv("nilai_mahasiswa.csv")

# 1. Hitung nilai statistik kehadiran
print("Rata-rata kehadiran:", df["kehadiran"].mean())
print("Minimum kehadiran  :", df["kehadiran"].min())
print("Maksimum kehadiran :", df["kehadiran"].max())

# 2. Buat Histogram
plt.hist(df["kehadiran"], bins=8, edgecolor="white")
plt.title("Histogram Kehadiran Mahasiswa")
plt.xlabel("Persentase Kehadiran")
plt.ylabel("Jumlah Mahasiswa")
plt.show()
```

```python
# Hitung nilai akhir sesuai bobot sebelumnya
df["nilai_akhir"] = 0.4 * df["uts"] + 0.6 * df["uas"]

# Tambahkan kolom status batas 60 (vektorisasi menggunakan apply)
df["status"] = df["nilai_akhir"].apply(lambda x: "LULUS" if x >= 60 else "REMIDI")

# Filter mahasiswa REMIDI
df_remidi = df[df["status"] == "REMIDI"]

# Tampilkan jumlah mahasiswa REMIDI per kelas
print("Jumlah Mahasiswa REMIDI per kelas:")
print(df_remidi.groupby("kelas")["nama"].count())
```
